Service Request Volume by Hour of Day

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd  # Make sure pandas is imported

# Extract hour from 'Created Date'
df['Hour'] = pd.to_datetime(df['Created Date']).dt.hour

# Count requests per hour, including all hours from 0 to 23
hourly = df['Hour'].value_counts().sort_index()
hourly = hourly.reindex(range(24), fill_value=0)  # Ensure all hours are present

# Plotting
plt.figure(figsize=(10, 4))
plt.plot(hourly.index, hourly.values, marker='o')
plt.title("Service Requests by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Requests")
plt.xticks(range(24))  # Ensure x-axis has ticks for all hours
plt.grid(True)
plt.tight_layout()
plt.show()

```

Distribution of Service Request Response Times (in Hours)

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd  # Ensure pandas is imported

# Calculate response time in hours
df['RequestClosingTime'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600
df = df[df['RequestClosingTime'] > 0]

# Define custom bin edges from 0 to 100 in steps of 10
bins = list(range(0, 110, 10))  # [0, 10, 20, ..., 100]

# Plot histogram with defined bins
plt.figure(figsize=(10, 4))
plt.hist(df['RequestClosingTime'], bins=bins, color='skyblue', edgecolor='black')
plt.title('Distribution of Response Times')
plt.xlabel('Response Time (Hours)')
plt.ylabel('Number of Requests')
plt.xticks(bins)  # Show bin edges on x-axis
plt.grid(True)
plt.tight_layout()
plt.show()
```

Complaint Type by Borough

```python
import matplotlib.pyplot as plt
import seaborn as sns

complaints_by_borough = df.groupby(['Borough', 'Complaint Type']).size().unstack().fillna(0)
top5 = df['Complaint Type'].value_counts().nlargest(5).index
complaints_by_borough[top5].plot(kind='bar', figsize=(12, 5))
plt.title("Top 5 Complaint Types by Borough")
plt.xlabel("Borough")
plt.ylabel("Number of Complaints")
plt.legend(title="Complaint Type")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate request closing time in hours
df['Request_Closing_Time'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600

# Filter out invalid values
df = df[df['Request_Closing_Time'] > 0]

# Get top 10 most frequent complaint types
top_10_complaints = df['Complaint Type'].value_counts().nlargest(10).index

#  Filter DataFrame to only include top 10 complaint types
df_top10 = df[df['Complaint Type'].isin(top_10_complaints)]

# Group by Complaint Type and Borough, calculate average response time
grouped = df_top10.groupby(['Complaint Type', 'Borough'])['Request_Closing_Time'].mean().unstack()

# Fill missing data
grouped = grouped.fillna(0)

# Plot as heatmap
plt.figure(figsize=(12, 5))
sns.heatmap(grouped, cmap='viridis', linewidths=0.5, annot=True, fmt=".1f")
plt.title('Avg Request Closing Time by Top 10 Complaint Types and Borough')
plt.xlabel('Borough')
plt.ylabel('Complaint Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

●Whether the average response time across complaint types is similar or not.

```python
from scipy.stats import f_oneway

# Create a list of closing times for each complaint type
groups = [group['Request_Closing_Time'].values for name, group in df.groupby('Complaint Type')]

# Perform ANOVA
f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("There is a significant association between Complaint Type and Borough.")
else:
    print("No significant association between Complaint Type and Borough.")
```

●Whether the type of complaint or service requested and location are related.

```python
from scipy.stats import chi2_contingency

# Create contingency table using Borough instead of Location Group
contingency_table = pd.crosstab(df['Complaint Type'], df['Borough'])

# Perform chi-square test of independence
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Printing results
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p)

alpha = 0.05
if p < alpha:
    print("There is a significant association between Complaint Type and Borough.")
else:
    print("No significant association between Complaint Type and Borough.")
```
